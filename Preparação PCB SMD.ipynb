## Preparação Circuito SMD ##

Este guia apresenta um roteiro passo a passo para a confecção de um circuito com predominância de componentes SMD (Surface Mount Device). Esses elementos são projetados para serem soldados diretamente na face da placa de circuito impresso (PCI). Por se tratar de uma operação complexa que exige elevado grau de exatidão, descreveremos, etapa a etapa, os procedimentos de usinagem da placa, incluindo tanto a fresagem quanto o corte assistido por laser. A finalidade é auxiliar na obtenção de resultados precisos e plenamente funcionais, abrangendo desde a geração e verificação dos arquivos de fabricação até o fresamento e o acabamento final. Siga as orientações abaixo para aprimorar e otimizar seu processo.

Assim como o roteiro de **Preparação do Layout do Circuito Impresso**, iremos desensolver o projeto simples para acender um LED com um interruptor, só que desta vez o circuito será totalmente SMD (Surface Mount Device), utilizamos uma tecnica da utilização da mascara de solda, (explicar como funciona beneficios para placa, circuito projeto, vida util, proteção).

### 2. Utilização do Inkscape

O inkscape é uma interface que vetoriza projetos para corte na CNC Laser (se tiver incoerente corriga e descreva melhor), que auxiliará no processo de raspagem dos pads, ondem serão aferidos os componentes no processo de soldagem. Confira as configurações:

1. Baixe e instale o Inkscape pelo site: https://inkscape.softonic.com.br/.

2. Com o projeto finalizado, exporte os arquivos SVG: Borda, multistrato e trilhas (se fosse um circuito totalmente PTH nao precisaria ser adicionado o arquivo de trilhas), para isso acesse **Arquivo > Exportar > SVG**.

3. Abra o arquivo SVG no Inkscape.

4. Na barra inferior à esquerda, clique no **"X"** e selecione a cor vermelha com o botão direito do mouse. Em seguida, clique em **Aplicar ao Contorno**.

6. Acesse **Preenchimento e Contorno > Estilo do Contorno** e configure em contorno mínimo.

7. Apague a linha mais externa da borda, selecionando-as e pressionando **Delete**.

8. Configure os pads (apresente neste circulo um sinonimo para a pessoa entenda do que se trata) para engrave, aplicando contorno e preenchimento no ícone cor preta com a tecla direita do mouse.

9. O circulo definido na origem deve ser envolvido por um quadrado para ser destacado, ambos devem esta em contorno minimo e configurados para corte (ícone cor vermelha).

10. Configure a borda da placa para corte (ícone cor vermelha).

11. Redimensione o desenho para ajustar às extremidades das trilhas em **Arquivo > Propriedades do Documento > Exibição > Redimensionar para o Conteúdo**, e salve o arquivo.



## Preparação da PCB na CNC Laser

Siga as sequintes orientações para iniciar o processo de fabricação:

1. Recorte a placa equivalente ao tamanho da forma.

2. Abra o arquivo no K40, software de inteface de conexão com a laser.

3. Fixe a moldura na mesa de trabalho do laser e aplique a placa, antes de aferi-la a revista com vinil.

4. Posicione o laser na placa, anotando as coordenadas de origem de trabalho.

5. Realize o vector cut 9mm/s com 9 % de potencia.

6. Retire a placa do molde, destacando o vinil, ate que somente a area do quadrado esteja coberta, destacando apenas o círculo interno para ser o ponto de origem de referência na minifresadora.

7. Mantenha a maquina ligada para aguardar a proxima etapa.
 



Se quiser saber o processo detalhado acesse o roteiro **Preparação Para o Processo de Corrosão > utilização do K40**.

### 3. Utilização Flatcam

O FlatCAM é um software utilizado para preparar arquivos de fresagem a partir do layout do circuito (arquivo Gerber). Ele permite definir os parâmetros necessários para o processo de usinagem, para o circuito em questao utilizaremos as mesmas configurções apresentadas no roteiro **Preparação Para o Processo de Usinagem**, exceto que nao ajustaremos o circuito para origem e nem faremos o desespelamento ja que o circuito esta na surperficie da placa.

Antes de fazer a preparaçao do gerber no flatcam, deve-se apagar o circulo de referencia antes das extrações dos arquivos. ## 4. Processo de Fresagem

Antes de iniciar o processo de fresagem revise os tópicos 2 e 3 do roteiro **Preparação Para o Processo de Usinagem**, para se familiarizar  e organizar as ferramentas que serão utilizadas, a seguir sigas os seguintes passos:


1. Fixe o outro molde na mesa de tralho da fresadora e anexe a placa que estava na laser.

2. Aplique a fresa v-bit ao spindle e a manusie ate a posição exata do circulo demarcado com o vinil.


3. Abra o arquivo de trilhas do projeto, no software candle, certifique-se que esta conectado com a maquina.

4. Realize o probe (medição em Z), no circulo em questão, lembre-se de fazer o processo com a ponta de prova de contato em contato fixo com o cobre da placa.

4. Em seguida, retire o vinil e  realize  mapa de altura.

5. finalize as configuraçoes e realize a fresagem das trilhas.

6. Retire a placa do molde, verique se isolou as trilhas e se a algum curto, limpea.

7. Mantenha a conecção da maquina parq que nao perca-se a referencia para proxima etapa. 


Observação: A limpeza da placa deve ser realizada com alcool isopropilico e palha de aço.

## 5. Processo de Curação da Tinta fotosensivel (Mascara de solda)

Neste processo utilizaremos a luz ultra violeta responsavel por curar a tinta fotosensivel (explicar com termos tecnicos), logo inciaremos o processo utizando os seguintes materiais:

- Folha de acetato transparente (tamanho um pouco maior do que a area da placa);
- Algum tipo de prença ou estura uniforme para disperção da mascara na placa;
- Uma fonte de luz Uv forte;
- Tinta fotosensivel

Siga os seguintes passos:

1. Transfira a tinta fotosensivel a placa, e a cubra com uma folha de acetato, faça a compressão uniforme com uma prença de modo a espalhar uniformemente a mascara na placa.

2. Em uma ambiente reservado coloque a placa na estufa de luz UV e deixe por aproximadamente 3 minutos, para que que cure a tinta.

3. Retire o acetato sobreposto a placa e leve novamente a placa de moldura que esta fixa na laser.

4. Realize o Raster Engrave 300 mm/s com 17% de potencia.

Observação: Aluz UV traz maleficios aos olhos, logo não deve observa-la.
