## Preparação de Circuito SMD ##

Este guia descreve a confecção de circuitos com componentes SMD (Surface Mount Device), soldados diretamente na superfície da placa de circuito impresso (PCI). O processo, que exige alta precisão, envolve fresagem, corte a laser, aplicação de máscara de solda e acabamento. A máscara de solda protege trilhas contra oxidação e curtos, melhora a soldagem e aumenta a durabilidade da placa.

Desenvolveremos um circuito simples para acender um LED com interruptor, totalmente em SMD, adaptando o roteiro **Preparação do Layout 
do Circuito Impresso** com componentes SMD. Veja o Vídeo 12: [link]. O resultado em 3D está na Figura 1.

### 1. Preparação dos Arquivos

No processo de fabricação com máscara de solda, a placa fica totalmente isolada, exceto nos pads (áreas de contato elétrico onde os componentes serão soldados). Para expor esses pads, utilizamos laser para remover a máscara de forma precisa. Como a fabricação envolve duas máquinas CNC (laser e minifresadora), é essencial definir um ponto de referência comum para alinhamento perfeito. Crie uma moldura padronizada para fixar a placa em ambas as máquinas, com dimensões suficientes para acomodar o circuito, sugerida na Figura 2.

**(FIGURA 2)**

Para definir essa referência no seu projeto finalizado no software EasyEDA:

- Arraste o projeto o mais próximo possível da origem (0,0).
- Adicione um furo ou marca circular de 0,5 mm na posição (0,0) para servir como ponto de referência.

### 2. Utilização do Inkscape

O Inkscape é um software de edição vetorial gratuito que prepara arquivos SVG para corte e gravação em CNC laser, facilitando a vetorização do layout para remoção precisa da máscara nos pads. 

**Recomendação**: No layout original, crie pads maiores que o necessário. No SVG, reduza ligeiramente sua área para centralizar a solda e compensar pequenas perdas de referência.

Siga estas configurações:

1. Se não possuir o software, baixe e instale o Inkscape no site oficial: https://inkscape.org/.

2. No software de layout, exporte os arquivos SVG: borda da placa, multicamadas e trilhas (essencial para circuitos SMD; em PTH, as trilhas podem ser omitidas). Acesse **Arquivo > Exportar > SVG**.

**(verificar um teste utilizando ja a camada de mascara de solda superior e bordas arquivo svg)**

3. Abra o arquivo SVG exportado no Inkscape.

4. Pressione Ctrl+A para selecionar toda a área de trabalho.

5. Acesse **Caminho > Converter Contorno em Caminhos**.

6. Na barra inferior esquerda, clique no ícone (X) e selecione a cor vermelha com o botão direito do mouse. Aplique ao contorno.

7. Acesse **Preenchimento e Contorno > Estilo do Contorno** e configure para contorno mínimo.

8. Delete a linha externa da borda e as trilhas, mantendo apenas os pads e a borda da placa.

9. Configure os pads (áreas circulares ou retangulares de soldagem) para gravação (engrave): aplique contorno e preenchimento em preto (clique direito no ícone de cor preta).

10. Envolva o círculo de referência na origem com um quadrado para destaque. Ambos devem ter contorno mínimo e serem configurados para corte (cor vermelha).

11. Configure a borda da placa para corte (cor vermelha).

10. Redimensione o documento para ajustar ao conteúdo: acesse **Arquivo > Propriedades do Documento > Exibição > Redimensionar** para o Conteúdo. Salve o arquivo.


observação: o ideal é que no layout do projeto voce deixe os pads bem maiores, e quando for fazer as alterações svg, deixe a area deste pad um pouco menos para a solda ser mais centralizada sem ter o risco de por causa de alguma perca de referencia pequena nao interfira na area de contato.

### 3.  Preparação da PCB na CNC Laser

Siga estas orientações para iniciar o processo de fabricação na laser:

1. Recorte a placa de fenolite no tamanho exato da moldura.

2. Abra o arquivo SVG preparado no software K40 (interface de controle da laser).

3. Fixe a moldura na mesa de trabalho da laser e insira a placa. Cubra a placa com vinil adesivo antes de fixá-la.

4. Alinhe o feixe do laser na placa e registre as coordenadas de origem.

5. Execute o corte vetorial (vector cut) a 9 mm/s com 9% de potência.

6. Retire a placa da moldura, remova o vinil exceto na área do quadrado de referência, e destaque apenas o círculo interno para marcar o ponto de origem na minifresadora.

7. Mantenha a máquina ligada para preservar as configurações para etapas posteriores.

Para detalhes mais aprofundados, consulte o roteiro **Preparação Para o Processo de Corrosão > Utilização do K40**.

### 4. Utilização Flatcam

O FlatCAM é um software open-source para preparar arquivos Gerber para fresagem CNC. Ele define parâmetros como profundidade de corte e caminhos de ferramenta. Para este circuito SMD, use as configurações do roteiro **Preparação Para o Processo de Usinagem**, exceto a parte de ajustar o circuito para a origem (mantenha o ponto de referência) e o desespelhamento, pois o circuito está na superfície.

Antes de processar:

- Apague o círculo de referência no layout.
- Exporte o arquivo Gerber da camada superior (TopLayer.GTL).

Siga o roteiro **Preparação Para o Processo de Usinagem > Utilização FlatCAM > Configuração das Camadas** e salve o G-code.

### 5. Processo de Fresagem

Antes de iniciar, revise os tópicos 2 e 3 do roteiro **Preparação Para o Processo de Usinagem** para organizar ferramentas e se familiarizar com o processo. Em seguida:

1. Fixe a outra moldura na mesa da minifresadora e insira a placa processada na laser.

2. Instale a fresa V-bit no spindle e posicione-a exatamente sobre o círculo demarcado com vinil.

3. Abra o G-code das trilhas no software Candle e confirme a conexão com a máquina.

4. Realize o probe em Z no círculo, usando a ponta de prova em contato com o cobre da placa.

5. Remova o vinil restante e execute o mapa de altura para compensar irregularidades.

6. Finalize as configurações e inicie a fresagem das trilhas.

7. Retire a placa, verifique isolamento das trilhas e ausência de curtos. Limpe com álcool isopropílico e palha de aço.


### 6. Processo de Curação da Máscara de Solda (Tinta Fotosensível)

A máscara de solda é uma tinta fotosensível que cura sob luz ultravioleta (UV), formando uma camada protetora polimerizada, garantindo adesão e durabilidade a PCB.
Materiais necessários:

- Folha de acetato transparente (maior que a placa);
- Espátula, rodo ou prensa para espalhar a tinta uniformemente;
- Fonte de luz UV (ex.: estufa UV);
- Tinta fotosensível para máscara de solda.

Siga os seguintes passos:

1. Transfira a tinta fotosensivel a placa, e a cubra com uma folha de acetato, faça a compressão uniforme com uma prença de modo a espalhar uniformemente a mascara na placa.

2. Em uma ambiente reservado coloque a placa na estufa de luz UV e deixe por aproximadamente 3 minutos, para que que cure a tinta.

3. Retire o acetato sobreposto a placa e leve novamente a placa de moldura que esta fixa na laser.

4. Realize o Raster Engrave 300 mm/s com 17% de potencia.

Observação: A luz UV traz maleficios aos olhos, logo não deve observa-la.

### 7. Soldagem dos Componentes

Para soldar componentes SMD com precisão, utilize pasta de solda e uma estação de refluxo ou forno. Um molde facilita a aplicação uniforme da pasta, mas sua criação é opcional. Se optar por usá-lo, reutilize o arquivo SVG criado anteriormente no Inkscape, ajustando apenas as configurações dos contornos dos pads para corte (cor vermelha). Prepare o molde da seguinte forma:

- Corte um molde de acetato com aberturas para os pads usando laser (12 mm/s, 10,5% de potência), com dimensões ligeiramente maiores que a borda da placa para facilitar o alinhamento.

Passos:

1. Se usar o molde, posicione-o sobre a placa, alinhando os pads, e aplique a pasta de solda uniformemente com uma espátula. Caso contrário, aplique a pasta diretamente nos pads com cuidado.

2. Coloque os componentes SMD nos pads com precisão, usando uma pinça se necessário.

3. Reflua a solda em um forno de refluxo ou com estação de ar quente, mantendo a temperatura entre 220-250°C (para solda sem chumbo).

4. Inspecione visualmente as juntas de solda e teste o circuito para verificar a funcionalidade.
